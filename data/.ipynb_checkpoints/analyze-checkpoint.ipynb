```python
import pandas as pd 
import numpy as np 
```

```python
df_kk_cn = pd.read_csv('input/kishakoutei_china.csv')
df_ar_cn = pd.read_csv('input/arxiv_china.csv')
df_ar_noncn = pd.read_csv('input/arxiv_nonchina.csv')
```

```python
pd.set_option('display.max_colwidth', None)
```

```python
df_kk_cn.head(10)
```

```python
df_kk_cn['title'].value_counts()
```

```python
papers_kk_cn = df_kk_cn['title'].unique()
```

```python
i=0
n=0
k=0
d=0
e=0
for paper in papers_kk_cn:
    print(paper)
    df_temp = df_kk_cn[df_kk_cn['title'] == paper]
    a = len(df_temp[(df_temp['language'] == 'cn')|(df_temp['language'] == 'zh')])
    b = len(df_temp[(df_temp['language'] == 'en')&((df_temp['countries'].str.contains('CN'))|(df_temp['countries'].str.contains('HK')))])
    c = len(df_temp)
    d = d+a
    e = e+b
    f = len(df_temp[(df_temp['language'] == 'en')&(df_temp['countries'].str.contains('US'))])
    
    print(a,b,c-b-a,f)
    print(round(a/c,2),round(b/c,2),round((c-b-a)/c,2))
    if a/c > 0.3:
        print ('yes')
        i=i+1
    if b/c > 0.3:
        print ('yes')
        n=n+1
    if (a+b)>f:
        print('china')
        k=k+1
    else:
        print('US')
print(i,n,k)
print(d/28,e/28,k/28)
```

```python
papers_ar_cn = df_ar_cn['citing_title'].unique()
```

```python
## a 中国語の数a
## b 中国の機関の数
## c 全体の数
## d e 中国の数の合計（平均用）
## f アメリカの機関の数
## g 中国かつアメリカの数
## i 中国語が1/3以上の数
## n 中国の機関が1/3以上の数
## k 中国の機関がアメリカの機関より多い論文の数
## l 同数の論文の数
```

```python
i=0
n=0
k=0
l=0
d=0
e=0
g=0
for paper in papers_ar_cn:
    print(paper)
    df_temp = df_ar_cn[df_ar_cn['citing_title'] == paper]
    a = len(df_temp[(df_temp['language'] == 'cn')|(df_temp['language'] == 'zh')])
    b = len(df_temp[(df_temp['language'] == 'en')&((df_temp['countries'].str.contains('CN'))|(df_temp['countries'].str.contains('HK')))])
    c = len(df_temp)
    d = d+a
    e = e+b
    f = len(df_temp[(df_temp['language'] == 'en')&(df_temp['countries'].str.contains('US'))])
    g= len(df_temp[(df_temp['language'] == 'en')&((df_temp['countries'].str.contains('CN'))|(df_temp['countries'].str.contains('HK')))&(df_temp['countries'].str.contains('US'))])
    
    print(a,b,c-b-a,f,g)
    print(round(a/c,2),round(b/c,2),round((c-b-a)/c,2))
    if a/c > 0.3:
        print ('yes')
        i=i+1
    if b/c > 0.3:
        print ('yes')
        n=n+1
    if (a+b)>f:
        print('china')
        k=k+1
    elif (a+b)==f:
        print('same')
        l=l+1
    else:
        print('US')
    df_temp2 =df_temp['countries'].value_counts()
    #print(df_temp2.index[0],df_temp2.iat[0])
    #print(df_temp2.index[1],df_temp2.iat[1])
    #print(df_temp2.index[2],df_temp2.iat[2])
    #print(df_temp2.index[3],df_temp2.iat[3])

    print("------")

    
print(i,n,k,l)
print(d/25,e/25,k/25)
```
