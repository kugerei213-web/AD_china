```python
from requests.exceptions import RequestException

if 'dataset' not in globals():
    dataset = []
if 'failed' not in globals():
    failed = []

done = {r['link'] for r in dataset}       
i = len(dataset)                            

def pick_en(vals):
    if len(vals) > 1:
        return vals[1]
    return vals[0] if vals else None

def pick_cn(vals):
    return vals[0] if vals else None

def fetch(url, headers, retries=3):
    for k in range(retries):
        try:
            return requests.get(url, headers=headers, timeout=30)
        except RequestException as e:
            wait = 5 * (2 ** k)             # 5秒 → 10秒 → 20秒
            print(f"  リトライ{k+1}/{retries}: {type(e).__name__} ({wait}秒待機)")
            time.sleep(wait)
    return None

for paper in data_total:
    url = paper['link']

    if url in done:                          # 再実行時は取得済みを飛ばす
        continue

    print(url)
    resp = fetch(url, headers)
    if resp is None or resp.status_code != 200:
        code = resp.status_code if resp is not None else "接続失敗"
        print("取得失敗:", code, url)
        failed.append(url)                   # 後でまとめて再取得
        continue

    resp.encoding = "utf-8"
    tree = etree.HTML(resp.text)

    xpath_title = '//meta[@name="citation_title"]/@content'
    elements_title = tree.xpath(xpath_title)
    xpath_name = '//meta[@name="citation_authors"]/@content'
    elements_name = tree.xpath(xpath_name)
    xpath_date = '//meta[@name="citation_date"]/@content'
    elements_date = tree.xpath(xpath_date)
    xpath_keywords = '//meta[@name="DC.Keywords"]/@content'
    elements_kewords = tree.xpath(xpath_keywords)
    xpath_abstract = 'string(//p[strong[contains(.,"Abstract")]]/following-sibling::p[1])'
    elements_abstract = tree.xpath(xpath_abstract)

    i = i + 1
    row = {
        'title':    pick_cn(elements_title),
        'title_en': pick_en(elements_title),
        'name':     pick_en(elements_name),
        'date':     elements_date[0] if elements_date else None,
        'keywords': pick_en(elements_kewords),
        'abstract': elements_abstract or None,
        'link':     url,
        'id':       i,
    }
    dataset.append(row)
    done.add(url)                            # 同一実行内の重複も防ぐ
    print(i)

    if len(dataset) % 200 == 0:              # 定期保存
        pd.DataFrame(dataset).to_json("dataset_tmp.jsonl", orient="records",
                                      lines=True, force_ascii=False)
        print(f"  --- 中間保存 {len(dataset)}件 ---")

    time.sleep(2)
```
