```python
LEAD = re.compile(
    r'^(?:'
    r'[\[［(（](\d{1,4})[\]］)）]'      # 括弧付き: 区切り不要
    r'|'
    r'(\d{1,3})[.．、\s]'              # 裸の数字: 区切り必須（誤検出防止）
    r')\s*(.*)$', re.S)

def to_pairs(text):
    if not isinstance(text, str) or not text.strip():
        return []
    if len(re.findall(r'\[\d+\]\s', text)) > 1:
        parts = re.split(r'(?=\[\d+\]\s)', text)
    else:
        parts = [text]
    out = []
    for p in parts:
        p = p.strip()
        if not p:
            continue
        m = LEAD.match(p)
        if m:
            no = m.group(1) or m.group(2)   # どちらの形式でマッチしたか
            out.append((no, m.group(3).strip()))
        else:
            out.append((None, p))
    return out

df['_p'] = df['reference'].apply(to_pairs)
df = df.explode('_p', ignore_index=True)
df[['ref_no', 'reference']] = pd.DataFrame(df['_p'].tolist(), index=df.index)
df = df.drop(columns='_p')

# 番号なし＝翻訳を除去
df_dropped = df[df['ref_no'].isna()]            # 確認用に残す
df = df[df['ref_no'].notna()].reset_index(drop=True)
df['ref_no'] = df['ref_no'].astype(int)
```

```python
MARK = re.compile(r'[\[［]([A-Z]{1,2})[\]］]')

def parse(s):
    if not isinstance(s, str):
        return (None, None, None, None)
    m = MARK.search(s)
    if not m:
        return (None, None, None, None)
    head = s[:m.start()]                       # 著者＋タイトル
    tail = s[m.end():].lstrip('.． ')           # 掲載誌以降
    # headの「最後の」ピリオドで分割。イニシャルのピリオドは著者側に残る
    parts = re.split(r'[.．](?!.*[.．])', head, maxsplit=1)
    if len(parts) == 2:
        return (parts[0].strip(), parts[1].strip(), m.group(1), tail.strip())
    return (None, head.strip(), m.group(1), tail.strip())

df_cn[['ref_author', 'ref_title', 'ref_type', 'ref_journal']] = pd.DataFrame(
    df_cn['reference'].apply(parse).tolist(), index=df_cn.index)
```

```python
def norm(s):
    if not isinstance(s, str):
        return None
    s = unicodedata.normalize("NFKC", s)   # 全角英数・全角記号を半角へ統一
    s = re.sub(r'\s+', '', s)              # 空白を全除去
    return s.strip('.．,，').lower()

corpus = set(df_cn['title'].map(norm).dropna())    # 収録論文のタイトル集合
df_ad_cn['ref_title_n'] = df_ad_cn['ref_title'].map(norm)
df_ad_cn['is_internal'] = df_ad_cn['ref_title_n'].isin(corpus)
```
